# 1 Problem Definition <a class='anchor' id='1'></a>

In the highly competitive financial services industry, **customer retention** is a critical factor that directly influences a bank’s profitability and long-term sustainability. Acquiring new customers is significantly more expensive than retaining existing ones, and studies have shown that a small increase in customer retention can lead to a substantial increase in profits. However, banks today face growing challenges in maintaining customer loyalty due to increasing service alternatives, digital transformation, and changing customer expectations.

**Customer churn**, defined as the loss of clients who stop using a bank's services, presents a major business problem. Identifying the early signs of churn can enable banks to implement targeted interventions and personalized retention strategies before customers actually leave. Hence, the ability to accurately **predict which customers are at risk of churning** is a valuable business capability.

This project focuses on developing a **data-driven churn prediction model** using machine learning and data mining techniques. The aim is to analyze historical customer behavior, demographic profiles, account activity, and service usage patterns to identify key predictors of churn. By training and evaluating models, we aim to **build an interpretable and accurate predictive system** that can be used by banking institutions to:

- Predict whether a customer is likely to churn in the near future,

- Understand which features (e.g., credit score, tenure, balance) contribute most to churn,

- Support strategic decisions for personalized marketing, customer engagement, and risk management.

The real-world relevance of this problem lies in its direct applicability to **customer relationship management (CRM)** in the banking sector. A successful solution will not only help reduce customer attrition but also improve customer satisfaction, loyalty, and lifetime value.

## 2.3 Handling Class Imbalance

```python
# check if the dataset is unbalanced or balanced
fig = plt.figure(figsize = (8,5))
df.Exited.value_counts(normalize = True).plot(kind='bar', color= ['skyblue','navy'], alpha = 0.9, rot=0)
plt.title('Exited Indicator No(0) and Yes(1) in the Imbalanced Dataset')
plt.show()
```

 As we can notice, target variable `Exited` is highly unbalanced, which means that further used ML models will tend to prefer target variable with value 0 over 1. It needs to be balanced in order to be able to implement various machine learning models with fair, unbiased results.

We will use oversampling strategy using `resample()`, thus we will generate same numbers of rows for the value 1 of target variable `Exited`, therefore we will get the same number of rows for both values of target variable.

```python
# use here the oversampling of the minority class
from sklearn.utils import resample

no = df[df.Exited == 0]
yes = df[df.Exited == 1]
yes_oversampled = resample(yes, replace=True, n_samples=len(no), random_state=42)
oversampled = pd.concat([no, yes_oversampled])
print(oversampled.shape)

fig = plt.figure(figsize = (8,5))
oversampled.Exited.value_counts(normalize = True).plot(kind='bar', color= ['skyblue','navy'], alpha = 0.9, rot=0)
plt.title('Exited Indicator No(0) and Yes(1) after Oversampling (Balanced Dataset)')
plt.show()
```

Then we drop `RowNumber`, `CustomerId` and `Surname` variables in order to train a model to understand general patterns.

Then we transform categorical attributes into numerical using `LabelEncoder()` in order to be able to provide these attributes as input for machine learning models.

```python
oversampled.select_dtypes(include=['object']).columns
```

```python
# Convert categorical features to continuous features with Label Encoding
from sklearn.preprocessing import LabelEncoder
lencoders = {}
for col in oversampled.select_dtypes(include=['object']).columns:
    lencoders[col] = LabelEncoder()
    oversampled[col] = lencoders[col].fit_transform(oversampled[col])
```

Now we detect and eliminate outliers from the inter-quartile interval-based data set.

```python
# Detecting outliers with IQR
Q1 = oversampled.quantile(0.25)
Q3 = oversampled.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
# Removing outliers from the dataset
oversampled = oversampled[~((oversampled < (Q1 - 1.5 * IQR)) |(oversampled > (Q3 + 1.5 * IQR))).any(axis=1)]
oversampled.shape
```

We observe that the data before removing outliers had 15926 rows. After running a code snippet for removing outliers, the dataset now has 15349 rows. As a result, the dataset is now free of 577 outliers.

## 3.1 Logistic Regression

### 3.1.1 Model Implementation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    roc_curve, roc_auc_score, cohen_kappa_score
)

# Partition and standardize data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_scaled = scaler.fit_transform(X)

# Hyperparameter tuning 
param_grid = {'C': [0.01, 0.1, 1, 10, 100]}
grid = GridSearchCV(LogisticRegression(max_iter=1000, random_state=42),
                    param_grid, cv=5, scoring='accuracy')
grid.fit(X_train_scaled, y_train)
best_model = grid.best_estimator_

print("\nOptimal Parameter：", grid.best_params_)
```

### 3.1.2 Evaluation and Analysis

```python
import warnings
warnings.filterwarnings('ignore')  # 忽略调优时的一些不必要警告
# model evaluation
y_pred = best_model.predict(X_test_scaled)
y_proba = best_model.predict_proba(X_test_scaled)[:, 1]
print("\nModel Evaluation:")
print("Accuracy:", accuracy_score(y_test, y_pred))
# Cross-Validation
cv_scores = cross_val_score(best_model, X_scaled, y, cv=5)
print("Cross-Validation Accuracy (5-fold):", np.mean(cv_scores))
print("Cohen's Kappa:", cohen_kappa_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
# Visualize the confusion matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.tight_layout()
plt.show()

# ROC and AUC
fpr, tpr, _ = roc_curve(y_test, y_proba)
roc_auc = roc_auc_score(y_test, y_proba)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.tight_layout()
plt.show()

# Feature importance values (logistic regression coefficients)
importance = np.abs(best_model.coef_[0])
feature_names = X.columns

importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
}).sort_values(by='Importance', ascending=False)

print("\nFeature importance values：")
print(importance_df)

# Visualize feature importance
plt.figure(figsize=(10, 6))
sns.barplot(data=importance_df, x='Importance', y='Feature', palette='coolwarm')
plt.title('Logistic Regression Feature Importance (|Coefficient|)')
plt.tight_layout()
plt.show()

# Histogram of predicted probabilities
plt.figure(figsize=(8, 5))
sns.histplot(y_proba, bins=20, kde=True, color='steelblue')
plt.title('Predicted Probabilities Distribution')
plt.xlabel('Predicted Probability (Positive Class)')
plt.ylabel('Frequency')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 3.2 Decision Tree

### 3.2.1 Model Implementation

```python
import os
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_graphviz, export_text, plot_tree
from sklearn.model_selection import train_test_split, GridSearchCV

# Split training and test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Hyperparameter search space
params = {
    'max_depth': [3, 5, 7, 10],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5]
}

# ID3 (using entropy)
id3_model = DecisionTreeClassifier(criterion='entropy', random_state=42)
id3_grid = GridSearchCV(id3_model, params, cv=5, scoring='accuracy')
id3_grid.fit(X_train, y_train)
id3_best = id3_grid.best_estimator_
print("ID3 最佳参数：", id3_grid.best_params_)

# C4.5 (also uses entropy, sets the pruning parameter)
c45_model = DecisionTreeClassifier(criterion='entropy', ccp_alpha=0.01, random_state=42)
c45_grid = GridSearchCV(c45_model, params, cv=5, scoring='accuracy')
c45_grid.fit(X_train, y_train)
c45_best = c45_grid.best_estimator_
print("C4.5 最佳参数：", c45_grid.best_params_)

# CART (with gini)
gini_model = DecisionTreeClassifier(criterion='gini', random_state=42)
gini_grid = GridSearchCV(gini_model, params, cv=5, scoring='accuracy')
gini_grid.fit(X_train, y_train)
gini_best = gini_grid.best_estimator_
print("CART(Gini) 最佳参数：", gini_grid.best_params_)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# 可视化 ID3 决策树
plt.figure(figsize=(25, 12)) 
plot_tree(id3_best, feature_names=X.columns, class_names=['No', 'Yes'], filled=True, rounded=True, max_depth=3)
plt.title("ID3 Decision Tree (criterion='entropy')")
plt.show()

# 可视化 C4.5 决策树
plt.figure(figsize=(20, 10))
plot_tree(c45_best, feature_names=X.columns, class_names=['No', 'Yes'], filled=True, rounded=True)
plt.title("C4.5 Decision Tree (entropy + pruning)")
plt.show()

# 可视化 CART 决策树
plt.figure(figsize=(20, 10))
plot_tree(gini_best, feature_names=X.columns, class_names=['No', 'Yes'], filled=True, rounded=True, max_depth=3)
plt.title("CART Decision Tree (criterion='gini')")
plt.show()
```

### 3.2.2 Evaluation and Analysis

```python
from sklearn.metrics import accuracy_score, cohen_kappa_score, classification_report, confusion_matrix, roc_auc_score, roc_curve
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings('ignore')  


def evaluate_model(model, X_test, y_test, name):
    print(f"\n=== {name} ===")
    
    # Predict labels and probabilities
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:,1]
    
    # Accuracy
    acc = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {acc:.4f}")
    
    # Cohen's Kappa
    kappa = cohen_kappa_score(y_test, y_pred)
    print(f"Cohen's Kappa: {kappa:.4f}")
    
    # Classification report
    print("Classification Report:\n", classification_report(y_test, y_pred))
    
    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'{name} Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
    
    # ROC and AUC
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    auc = roc_auc_score(y_test, y_proba)
    print(f"ROC AUC: {auc:.4f}")
    plt.figure()
    plt.plot(fpr, tpr, label=f'{name} (AUC = {auc:.2f})')
    plt.plot([0,1], [0,1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{name} ROC Curve')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    # Feature Importances
    importances = model.feature_importances_
    feature_names = X_test.columns
    fi_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances})
    fi_df = fi_df.sort_values(by='Importance', ascending=False)
    print("Feature Importances:\n", fi_df)
    
    plt.figure(figsize=(8, 5))
    sns.barplot(x='Importance', y='Feature', data=fi_df, palette='coolwarm')
    plt.title(f'{name} Feature Importances')
    plt.tight_layout()
    plt.show()
    
    # Histogram of predicted probabilities
    plt.figure()
    sns.histplot(y_proba, bins=20, kde=True)
    plt.title(f'{name} Predicted Probabilities')
    plt.xlabel('Predicted Probability')
    plt.ylabel('Count')
    plt.grid(True)
    plt.show()

from sklearn.model_selection import cross_val_score

# Cross-Validation
print("Cross-Validation Scores:")
for model, name in zip([id3_best, c45_best, gini_best], ['ID3', 'C4.5', 'CART']):
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='accuracy')
    print(f"{name} CV Accuracy: {scores.mean():.4f} (+/- {scores.std():.4f})")

# Evaluate each model individually
evaluate_model(id3_best, X_test, y_test, 'ID3')
evaluate_model(c45_best, X_test, y_test, 'C4.5')
evaluate_model(gini_best, X_test, y_test, 'CART')
```

## 3.5 Artificial Neural Network

### 3.5.1 Model Implementation

```python
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import roc_auc_score, make_scorer
from skorch import NeuralNetClassifier
import warnings
warnings.filterwarnings('ignore')
from skorch.callbacks import EpochScoring

# 数据预处理
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

selector = SelectFromModel(LogisticRegression(max_iter=1000), threshold='mean')
X_selected = selector.fit_transform(X_scaled, y)

X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.2, random_state=42)

# 构建 PyTorch 模型
class ChurnNet(nn.Module):
    def __init__(self, input_dim=10, hidden1=64, hidden2=32):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden1)
        self.fc2 = nn.Linear(hidden1, hidden2)
        self.out = nn.Linear(hidden2, 1)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        x = self.sigmoid(self.out(x))
        return x

# 使用 skorch 包装 PyTorch 模型
net = NeuralNetClassifier(
    module=ChurnNet,
    module__input_dim=X_train.shape[1],
    max_epochs=50,
    lr=0.01,
    optimizer=torch.optim.Adam,
    criterion=nn.BCELoss,
    iterator_train__shuffle=True,
    verbose=0,
    callbacks=[
        EpochScoring(scoring='accuracy', name='train_accuracy', on_train=True),
        EpochScoring(scoring='accuracy', name='valid_accuracy', on_train=False),
    ]
)

# 自定义评分指标：AUC（适用于不平衡数据）
auc_scorer = make_scorer(roc_auc_score, needs_proba=True)

# 参数网格
params = {
    'lr': [0.001, 0.005, 0.01],
    'optimizer': [torch.optim.Adam, torch.optim.SGD],
    'module__hidden1': [64, 128],
    'module__hidden2': [32, 64],
    'max_epochs': [50, 100],
}

# 网格搜索
gs = GridSearchCV(net, params, scoring=auc_scorer, cv=3, n_jobs=-1)
gs.fit(X_train.astype(np.float32), y_train.values.reshape(-1, 1).astype(np.float32))

# 输出最佳参数与得分
print("Best AUC: {:.4f}".format(gs.best_score_))
print("Best Parameters:")
print(gs.best_params_)
```
